```python
import pandas as pd
import json
```

```python
df = pd.read_csv("ev_us_location.csv")
```

```python
df["_doc"] = df["_doc"].apply(json.loads)
```

### _doc ko JSON me convert karo

```python
df["_doc"]
```

### Nested JSON ko flat columns me convert karo

```python
result = pd.json_normalize(df["_doc"])
```

```python
result
```

### Op aur outer _id ko add karo

```python

result.insert(0, "Op", df["Op"])
result.insert(1, "_id", df["_id"])
```
